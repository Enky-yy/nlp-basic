```python
import pandas
import numpy
```

```python
import os
import gensim
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
story = []
for filename in os.listdir('game-of-thrones-books'):
    f =open(os.path.join('game-of-thrones-books', filename), encoding='cp1252')
    corpus= f.read()
    raw_data = sent_tokenize(corpus)
    for sent in raw_data:
        sent = simple_preprocess(sent)
        story.append(sent)
```

```python
story
```

```python
len(story)
```

```python
model = gensim.models.Word2Vec(story, min_count=2, window=10,workers=4)
```

```python
model.build_vocab(story)
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
model.wv.most_similar('daenerys')
```
